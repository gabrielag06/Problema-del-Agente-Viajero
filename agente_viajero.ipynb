```python
import numpy as np
import time

MatrizDist = np.array([
    [99, 12, 10, 99, 99, 99, 12],
    [12, 99,  8, 12, 99, 99, 99],
    [10,  8, 99, 11,  3, 99,  9],
    [99, 12, 11, 99, 11, 10, 99],
    [99, 99,  3, 11, 99,  6,  7],
    [99, 99, 99, 10,  6, 99,  9],
    [12, 99,  9, 99,  7,  9, 99]
])

inicio = time.perf_counter()

nodoInicial = 1
rutaVisita = [1] # ruta que inicia en la ciudad 1
fallos = [] # registra si se queda atascado 
distanciaRecorrida = 0 # registra la distnacia total de la ruta (función objetivo)

iteraciones = 0 # contador de iteraciones
comparaciones = 0
while len(rutaVisita) < 7 or MatrizDist[rutaVisita[-1]-1][nodoInicial-1] == 99:
  iteraciones += 1
  nodoActual = rutaVisita[-1]
  mejorVecino = 0
  menorDistancia = 99

  for i in range(1, 8):
    comparaciones += 1
    if i not in rutaVisita and MatrizDist[nodoActual-1][i-1] < menorDistancia and (nodoActual, i) not in fallos:
      mejorVecino = i
      menorDistancia = MatrizDist[nodoActual-1][i-1]

  if mejorVecino != 0:
    rutaVisita.append(mejorVecino)
    distanciaRecorrida += menorDistancia
  else:
    nodoFallido = rutaVisita.pop()
    nodoAnterior = rutaVisita[-1]

    distanciaRecorrida -= MatrizDist[nodoAnterior-1][nodoFallido-1]
    fallos.append((nodoAnterior, nodoFallido))

    # reset de fallos
    fallos = [par for par in fallos if par[0] != nodoFallido]
  
iteraciones += 1
nodoActual = rutaVisita[-1]
distanciaRecorrida += MatrizDist[nodoActual-1][nodoInicial-1]
rutaVisita.append(nodoInicial)

fin = time.perf_counter()
tiempoEjecucion = fin - inicio

#print(fallos)
print("Ruta de visita:", rutaVisita)
print("Distancia total:", distanciaRecorrida)
print("Número de iteraciones:", iteraciones)
print("Evaluaciones internas:", comparaciones)
print("Tiempo de ejecución:", tiempoEjecucion, "segundos")
```
